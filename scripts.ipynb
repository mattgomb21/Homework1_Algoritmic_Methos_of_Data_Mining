5) Find a string

```python
def count_substring(string, sub_string):
    if len(string)<1 or len(string)>200 or not string.isascii() or not sub_string.isascii():
        print("Input non ammissibile")
    else:
        contatore=0
        for i in range(len(string)-len(sub_string)+1):
            string_corrente=string[i:i+len(sub_string)]
            if string_corrente==sub_string:
                contatore+=1
        return contatore


if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()

    count = count_substring(string, sub_string)
    print(count)
```

11) Alphabet Rangoli

```python
def print_rangoli(size):
    if 0<size<27:
        alfabeto="abcdefghijklmnopqrstuvwxyz"
        riga_centrale1=[alfabeto[size-j-1] for j in range(size)]
        riga_centrale="-".join(riga_centrale1+riga_centrale1[:-1][::-1]) #escludo l'ultimo elemento per non ripetere a due volte e inverto
        lunghezza=len(riga_centrale)
        righe=[]
        for i in range(1,size):
            sinistra=[alfabeto[size-j-1] for j in range(i)]
            lista=sinistra+sinistra[:-1][::-1]
            riga="-".join(lista).center(lunghezza,'-')
            righe.append(riga)
        for riga in righe:  #stampo parte superiore
            print(riga)
        print(riga_centrale)
        for riga in reversed(righe): #inverto lista con le righe per ottenere parte inferiore ribaltata
            print(riga)
    else:
        return("Input non ammissibile")

if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

13) The Minion Game

```python
def minion_game(string):
    if string.isupper() and 0<len(string)<=10**6:
        vocali="AEIOU"
        #Stuart per consonanti e Kevin per vocali
        punteggio_S=0
        punteggio_K=0
        lunghezza=len(string)
        for i in range(lunghezza):
            if string[i] in vocali:
                punteggio_K+=(lunghezza-i) #non creo tutte le sottostringhe (sarebbe troppo costoso) ma conto solamente il numero di parole possibili (ovvero tutte quelle che posso creare che iniziano con tale lettera)
            else:
                punteggio_S+=(lunghezza-i)
        if punteggio_S>punteggio_K:
            print("Stuart "+str(punteggio_S))
        elif punteggio_S<punteggio_K:
            print("Kevin "+str(punteggio_K))
        else:
            print("Draw")
    else:
        print("Input non ammissibile per il Minion Game")

if __name__ == '__main__':
    s = input()
    minion_game(s)
```

8) Piling Up!

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
from collections import deque

if __name__=='__main__':
    T=int(input())  #numero di test
    for _ in range(T):
        n=int(input())
        lunghezza_cubi=deque(map(int,input().split()))  #posso prendere solo il primo o l'ultimo elemento
        blocco_precedente=max(lunghezza_cubi[0],lunghezza_cubi[-1])
        possibile=True
        while lunghezza_cubi:
            if lunghezza_cubi[0]>=lunghezza_cubi[-1]:
                blocco_scelto=lunghezza_cubi.popleft()
            else:
                blocco_scelto=lunghezza_cubi.pop()
            if blocco_scelto>blocco_precedente:
                possibile=False
                break
            blocco_precedente=blocco_scelto
        if possibile:
            print("Yes")
        else:
            print("No")
```

2) XML2 - Find the Maximum Depth

```python
import xml.etree.ElementTree as etree

maxdepth = 0
def depth(elem, level):
    global maxdepth
    level+=1
    maxdepth=max(maxdepth,level)
    for x in elem:
        depth(x,level)

if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

4) Recursive Digit Sum

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'superDigit' function below.
#
# The function is expected to return an INTEGER.
# The function accepts following parameters:
#  1. STRING n
#  2. INTEGER k
#

def superDigit(n, k):
    # Write your code here
    if len(n)==1 and k==1:   #passo base
        return int(n)
    somma=0
    for x in n:
        somma+=int(x)
    somma=somma*k  #per evitare di concatenare subito la stringa (potrebbe dare problemi in memoria)
    return superDigit(str(somma),1)

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    first_multiple_input = input().rstrip().split()

    n = first_multiple_input[0]

    k = int(first_multiple_input[1])

    result = superDigit(n, k)

    fptr.write(str(result) + '\n')

    fptr.close()
```

5) Insertion Sort - Part 1

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'insertionSort1' function below.
#
# The function accepts following parameters:
#  1. INTEGER n
#  2. INTEGER_ARRAY arr
#

def insertionSort1(n, arr):
    # Write your code here
    def stampa(a):
        lista=[]
        for x in a:
            lista.append(str(x))
        print(" ".join(lista))
    valore=arr[n-1]
    i=n-2
    while i>=0 and arr[i]>valore:
        arr[i+1]=arr[i]
        stampa(arr)
        i=i-1
    arr[i+1]=valore
    stampa(arr)

if __name__ == '__main__':
    n = int(input().strip())

    arr = list(map(int, input().rstrip().split()))

    insertionSort1(n, arr)
```

6) Insertion Sort - Part 2

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'insertionSort2' function below.
#
# The function accepts following parameters:
#  1. INTEGER n
#  2. INTEGER_ARRAY arr
#

def insertionSort2(n, arr):
    # Write your code here
    def stampa(a):
        lista=[]
        for x in a:
            lista.append(str(x))
        print(" ".join(lista))
    valore=arr[n-1]
    for j in range(1,n):  #il primo elemento da solo gia ordinato quindi parto da indice 2
        valore=arr[j]
        i=j-1
        while i>=0 and arr[i]>valore:
            arr[i+1]=arr[i]
            i=i-1
        arr[i+1]=valore
        stampa(arr)

if __name__ == '__main__':
    n = int(input().strip())

    arr = list(map(int, input().rstrip().split()))

    insertionSort2(n, arr)
```
